BLOCO DE TRATAMENTO DOS CSV:

Objetivo:
- ADEQUAÇÃO DE NOMES 
- COLUNAS
- PADRONIZAÇÃO DE MAIÚSCULAS/MINÚSCULAS
- REMOVER ESPAÇOS VAZIOS
- VERIFICAÇÃO DE DUPLICATAS E REMOÇÃO DELAS.

```python
import pandas as pd

# ============================================================
# 1. TRATAMENTO DO CSV PRINCIPAL DE VENDAS 2024 
# ============================================================

#VERSÃO CASA:
#df_primarysales_2024_original = pd.read_csv(
  # r"C:\Users\dsz_d\Documents\Estudos\SENAC\BIGDATA\SenacAnaliseDados2026.2"
   #r"\UC02\Projeto_final_UC2\Fonte_Dados_Original\vgchartz-2024.csv"
#)

#VERSÃO CURSO: 
df_primarysales_2024_original = pd.read_csv(
   r"C:\\Users\\douglas.zanella\\Documents\\BIGDATA2026\\SenacAnaliseDados2026.2"
   r"\\UC02\\Projeto_final_UC2\\Fonte_Dados_Original\\vgchartz-2024.csv"
)

'''
OBSERVAÇÃO:
erros de caminho aconteciam o tempo inteiro, então em algumas pesquisar foi indicado usar:
"r" antes de um caminho para o python ler como RAW STRING (caminho bruto) 
para nao haver erros de interpretação comoo \t ou \n

'''
# Cria uma cópia para preservar o DataFrame original
df_primarysales_2024_tratado = (
    df_primarysales_2024_original.copy()
)

# Remove a coluna de imagens
df_primarysales_2024_tratado = (
    df_primarysales_2024_tratado.drop(
        columns=["img"]
    )
)

# Renomeia as colunas
df_primarysales_2024_tratado = (
    df_primarysales_2024_tratado.rename(
        columns={
            "title": "game_name",
            "console": "console_code",
            "release_date": "game_release_date",
            "last_update": "last_update_date"
        }
    )
)

#Renomeia os dados da coluna inteiro para padronizar: 
df_primarysales_2024_tratado["genre"] = df_primarysales_2024_tratado["genre"].replace('Role-Playing', 'RPG')

print("\nCSV PRINCIPAL DE VENDAS 2024:")
print(df_primarysales_2024_tratado.head())

print("\nQuantidade de linhas e colunas:")
print(df_primarysales_2024_tratado.shape)

print("\n", "=" * 80)


# ============================================================
# 2. TRATAMENTO DO CSV SECUNDÁRIO DE VENDAS
# ============================================================


# #VERSÃO CASA
# df_secundarysales_original = pd.read_csv(
#     r"C:\Users\dsz_d\Documents\Estudos\SENAC\BIGDATA\SenacAnaliseDados2026.2"
#     r"\UC02\Projeto_final_UC2\Fonte_Dados_Original\video_games_sales.csv"
# )

#VERSÃO CURSO:
df_secundarysales_original = pd.read_csv(
    r"C:\\Users\\douglas.zanella\\Documents\\BIGDATA2026\\SenacAnaliseDados2026.2"
    r"\\UC02\\Projeto_final_UC2\\Fonte_Dados_Original\\video_games_sales.csv"
)

# Cria uma cópia para preservar o DataFrame original
df_secundarysales_tratado = (
    df_secundarysales_original.copy()
)

# Tratamento maiusculo -> minusculo 
df_secundarysales_tratado.columns = (
    df_secundarysales_tratado.columns.str.lower()
)

# Renomeia colunas para padronizar com o projeto
df_secundarysales_tratado = (
    df_secundarysales_tratado.rename(
        columns={
            "rank": "game_rank",
            "name": "game_name",
            "platform": "console_code",
            "year": "game_year",
            "global_sales": "total_sales"
        }
    )
)

#Renomeia os dados da coluna inteiro para padronizar: 
df_secundarysales_tratado["genre"] = df_secundarysales_tratado["genre"].replace('Role-Playing', 'RPG')


print("\nCSV SECUNDÁRIO DE VENDAS:")
print(df_secundarysales_tratado.head())

print("\nQuantidade de linhas e colunas:")
print(df_secundarysales_tratado.shape)

print("\n", "=" * 80)

# ============================================================
# 3. TRATAMENTO DO NOVO CSV DE CONSOLES
# ============================================================

# #VERSÃO CASA
# df_consoles_original = pd.read_csv(
#     r"C:\Users\dsz_d\Documents\Estudos\SENAC\BIGDATA"
#     r"\SenacAnaliseDados2026.2\UC02\Projeto_final_UC2"
#     r"\Fonte_Dados_Original\videogameconsoles.csv",
#     encoding="cp1252",
#     skipinitialspace=True
# )


#VERSÃO CURSO:
df_consoles_original = pd.read_csv(
    r"C:\\Users\\douglas.zanella\\Documents\\BIGDATA2026"
    r"\\SenacAnaliseDados2026.2\\UC02\\Projeto_final_UC2"
    r"\\Fonte_Dados_Original\\videogameconsoles.csv",
    encoding="cp1252",
    skipinitialspace=True
)

# Cria uma cópia do DataFrame original
df_consoles_tratado = df_consoles_original.copy()


# PADRONIZA OS CABEÇALHOS
df_consoles_tratado.columns = (
    df_consoles_tratado.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)


print("\nCabeçalhos depois da padronização:")
print(df_consoles_tratado.columns.tolist())

# REMOVE A COLUNA QUE NÃO SERÁ UTILIZADA
if (
    "2022_price_adjusted_for_inflation"
    in df_consoles_tratado.columns
):

    df_consoles_tratado = df_consoles_tratado.drop(
        columns=[
            "2022_price_adjusted_for_inflation"
        ]
    )


# RENOMEIA AS COLUNAS
df_consoles_tratado = df_consoles_tratado.rename(
    columns={
        "primary_consoles": "console_name",
        "year_of_release": "console_year"
    }
)


print("\nCabeçalhos depois da remoção e renomeação:")
print(df_consoles_tratado.columns.tolist())


# ============================================================
# 3.1 CRIA console_code
# ============================================================

codigos_consoles = {
    # Primeiras gerações
    "Magnavox Odyssey": "ODY",
    "Home Pong": "PONG",
    "Coleco Telstar series": "TELSTAR",
    "Fairchild Channel F": "CHF",
    "Atari 2600": "2600",
    "Mattel Intellivision": "INT",
    "Magnavox Odyssey 2": "ODY2",
    "ColecoVision": "CV",

    # Terceira geração
    "Nintendo Entertainment System": "NES",
    "Sega Master System": "MS",
    "Atari 7800": "7800",

    # Quarta geração
    "TurboGrafx-16": "TG16",
    "Sega Genesis": "GEN",
    "Neo Geo": "NG",
    "Super Nintendo Entertainment System": "SNES",
    "Game Gear": "GG",
    "Game Boy": "GB",

    # Quinta geração
    "Sega Saturn": "SAT",
    "Sony PlayStation": "PS",
    "Game Boy Color": "GBC",
    "Nintendo 64": "N64",

    # Sexta geração
    "Sega Dreamcast": "DC",
    "Sony PlayStation 2": "PS2",
    "Nintendo GameCube": "GC",
    "Game Boy Advance": "GBA",
    "Microsoft Xbox": "XB",

    # Sétima geração
    "Microsoft Xbox 360": "X360",
    "Sony PlayStation 3": "PS3",
    "Nintendo DS": "DS",
    "Playstation Portable": "PSP",
    "Nintendo Wii": "WII",

    # Oitava geração
    "Nintendo Wii U": "WIIU",
    "Sony PlayStation 4": "PS4",
    "Microsoft Xbox One": "XONE",
    "Nintendo 3DS": "3DS",
    "Playstation Vita": "PSV",
    "Nintendo Switch": "NS",

    # Nona geração
    "Microsoft Xbox Series X": "XS",
    "Microsoft Xbox Series S": "XS",
    "Sony PlayStation 5 (Digital Edition)": "PS5",
    "Sony PlayStation 5 (Disc Edition)": "PS5"
}


df_consoles_tratado["console_code"] = (
    df_consoles_tratado["console_name"]
    .str.strip()
    .map(codigos_consoles)
)

# ============================================================
# 3.2 CRIA O FABRICANTE
# ============================================================

fabricantes_por_codigo = {
    "ODY": "Magnavox",
    "PONG": "Atari",
    "TELSTAR": "Coleco",
    "CHF": "Fairchild",
    "2600": "Atari",
    "INT": "Mattel",
    "ODY2": "Magnavox",
    "CV": "Coleco",

    "NES": "Nintendo",
    "MS": "Sega",
    "7800": "Atari",

    "TG16": "NEC",
    "GEN": "Sega",
    "NG": "SNK",
    "SNES": "Nintendo",
    "GG": "Sega",
    "GB": "Nintendo",

    "SAT": "Sega",
    "PS": "Sony",
    "GBC": "Nintendo",
    "N64": "Nintendo",

    "DC": "Sega",
    "PS2": "Sony",
    "GC": "Nintendo",
    "GBA": "Nintendo",
    "XB": "Microsoft",

    "X360": "Microsoft",
    "PS3": "Sony",
    "DS": "Nintendo",
    "PSP": "Sony",
    "WII": "Nintendo",

    "WIIU": "Nintendo",
    "PS4": "Sony",
    "XONE": "Microsoft",
    "3DS": "Nintendo",
    "PSV": "Sony",
    "NS": "Nintendo",

    "XS": "Microsoft",
    "PS5": "Sony"
}


df_consoles_tratado["manufacturer"] = (
    df_consoles_tratado["console_code"]
    .map(fabricantes_por_codigo)
)

# CONVERTE AS COLUNAS PARA TIPOS NUMÉRICOS
df_consoles_tratado["original_price"] = (
    df_consoles_tratado["original_price"]
    .astype("string")
    .str.strip()
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df_consoles_tratado["original_price"] = pd.to_numeric(
    df_consoles_tratado["original_price"],
    errors="coerce"
)


df_consoles_tratado["total_systems_sold"] = (
    df_consoles_tratado["total_systems_sold"]
    .astype("string")
    .str.strip()
    .str.replace(",", "", regex=False)
)

df_consoles_tratado["total_systems_sold"] = pd.to_numeric(
    df_consoles_tratado["total_systems_sold"],
    errors="coerce"
)

# Garante que o ano seja numérico
df_consoles_tratado["console_year"] = pd.to_numeric(
    df_consoles_tratado["console_year"],
    errors="coerce"
)


# ============================================================
# 3.3 VERIFICA CONSOLES SEM CONSOLE_CODE
# ============================================================

consoles_sem_codigo = df_consoles_tratado[
    df_consoles_tratado["console_code"].isnull()
]


print("\nConsoles sem código:")
print(
    consoles_sem_codigo[
        [
            "console_name",
            "generation",
            "console_year"
        ]
    ]
)

#  VERIFICAÇÃO DE DUPLICATAS
'''
 GARANTE QUE NÃO HAJA CODIGOS REPETIDOS PARA EVITAR ERROS NO SQL
 Xbox Series X e Series S possuem o mesmo código XS.
 As duas versões do PlayStation 5 possuem o mesmo código PS5.
'''
df_consoles_sql = (
    df_consoles_tratado
    .dropna(
        subset=["console_code"]
    )
    .groupby(
        "console_code",
        as_index=False
    )
    .agg(
        {
            "console_name": "first",
            "manufacturer": "first",
            "generation": "first",
            "time_period": "first",
            "console_year": "min",
            "game_media": "first",
            "original_price": "min",
            "total_systems_sold": "max"
        }
    )
)

# ============================================================
# 3.4 CRIA O IDENTIFICADOR DOS CONSOLES (ID_CONSOLE)
# ============================================================

df_consoles_sql.insert(
    0,
    "id_console",
    range(
        1,
        len(df_consoles_sql) + 1
    )
)


# ORGANIZA A ORDEM DAS COLUNAS
df_consoles_sql = df_consoles_sql[
    [
        "id_console",
        "console_code",
        "console_name",
        "manufacturer",
        "generation",
        "time_period",
        "console_year",
        "game_media",
        "original_price",
        "total_systems_sold"
    ]
]

# ============================================================
# 4. VERIFICAÇÃO DE NOMES REPETIDOS NAS TABELAS 
# ============================================================

#PADRONIZAÇÃO DE NOMES PARA VERIFICAÇÃO DE DUPLICATAS:
#TABELA PRINCIPAL
df_primarysales_2024_tratado["game_name_key"] = (
    df_primarysales_2024_tratado["game_name"]
    .str.strip() #RETIRA OS ESPAÇOS VAZIOS
    .str.lower() #TRANSFORMA EM MINÚSCULAS
)

df_primarysales_2024_tratado["console_code_key"] = (
    df_primarysales_2024_tratado["console_code"]
    .str.strip()
    .str.upper()
)

#TABELA SECUNDÁRIA
df_secundarysales_tratado["game_name_key"] = (
    df_secundarysales_tratado["game_name"]
    .str.strip()
    .str.lower()
)

df_secundarysales_tratado["console_code_key"] = (
    df_secundarysales_tratado["console_code"]
    .str.strip()
    .str.upper()
)

# ============================================================
# 5. CHECAGEM E TRATAMENTO DAS DUPLICATAS
# ============================================================

#PROCURA POR DUPLICATAS NA TABELA PRINCIPAL:
duplicados_principal = (
    df_primarysales_2024_tratado
    .duplicated(
        subset=[
            "game_name_key",
            "console_code_key"
        ]
    )
    .sum()
)

duplicados_secundario = (
    df_secundarysales_tratado
    .duplicated(
        subset=[
            "game_name_key",
            "console_code_key"
        ]
    )
    .sum()
)

print("\nRegistros repetidos na tabela principal:")
print(duplicados_principal)
print("-"*50)
print("\nRegistros repetidos na tabela secundária:")
print(duplicados_secundario)

# ADEQUAÇÃO DA TABELA SECUNDÁRIA PARA QUE HAJA APENAS UMA LINHA PARA CADA JOGO 
df_secundaria_resumida = (
    df_secundarysales_tratado
    .groupby(
        [
            "game_name_key",
            "console_code_key"
        ],
        as_index=False
    )
    .agg(
        {
            "total_sales": "mean",
            "na_sales": "mean",
            "jp_sales": "mean",
            "other_sales": "mean"
        }
    )
)

# CONFIRMA QUE AS CHAVES DA TABELA NÃO ESTÃO REPETIDAS
duplicados_restantes = (
    df_secundaria_resumida
    .duplicated(
        subset=[
            "game_name_key",
            "console_code_key"
        ]
    )
    .sum()
)

print("\nDuplicações restantes na tabela secundária resumida:")
print(duplicados_restantes)

print("\nLinhas da secundária original:")
print(len(df_secundarysales_tratado))

print("\nLinhas da secundária resumida:")
print(len(df_secundaria_resumida))

print("\nDuplicações restantes:")
print(
    df_secundaria_resumida.duplicated(
        subset=[
            "game_name_key",
            "console_code_key"
        ]
    ).sum()
)

# ============================================================
# 6. UNIÃO DAS TABELAS SEM DUPLCIATAS COM MERGE
# ============================================================

df_vendas_geral = df_primarysales_2024_tratado.merge(
    df_secundaria_resumida,
    on=[
        "game_name_key",
        "console_code_key"
    ],
    how="left",
    suffixes=(
        "_principal",
        "_secundario"
    ),
    validate="many_to_one"
)

# CHECAGEM DAS LINHAS :
print("\nLinhas da tabela principal:")
print(len(df_primarysales_2024_tratado))

print("\nLinhas depois do novo merge:")
print(len(df_vendas_geral))
```

BLOCO DE MÉDIAS ENTRE COLUNAS DAS TABELAS PARA CRIAÇÃO DE UM ÚNICO CSV:

Objetivo: Faz as médias entre valores das duas tabelas (vendas principal e Vendas Secundárias) afim de criar uma nova tabela com valores bem preenchidos e estruturados para utilizar nas análises

```python
# ============================================================
# 7. MÉDIA ENTRE OS VALORES DAS TABELAS PARA ADEQUAÇÃO:
# ============================================================

# CHECAGEM DE COLUNAS APÓS O MERGE:
print("\nColunas disponíveis depois do merge:")

for coluna in df_vendas_geral.columns:
    print(coluna)

# MÉDIA DE VENDAS GLOBAIS:
df_vendas_geral["total_sales_final"] = (
    df_vendas_geral[
        [
            "total_sales_principal",
            "total_sales_secundario"
        ]
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# VENDAS NA AMÉRICA DO NORTE:
df_vendas_geral["na_sales_final"] = (
    df_vendas_geral[
        [
            "na_sales_principal",
            "na_sales_secundario"
        ]
    ]
    .mean(
        axis=1,
        skipna=True
    )
) 

# VENDAS NO JAPÃO:
df_vendas_geral["jp_sales_final"] = (
    df_vendas_geral[
        [
            "jp_sales_principal",
            "jp_sales_secundario"
        ]
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# OUTRAS REGIÕES:
df_vendas_geral["other_sales_final"] = (
    df_vendas_geral[
        [
            "other_sales_principal",
            "other_sales_secundario"
        ]
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# ======= RESULTADO DAS MÉDIAS ==========
print(
    df_vendas_geral[
        [
            "game_name",
            "console_code",
            "total_sales_principal",
            "total_sales_secundario",
            "total_sales_final"
        ]
    ].head(20)
)

# CHECAGEM DE VALORES AUSENTES: 
print("\nValores ausentes  ANTES: ")

print(
    df_primarysales_2024_tratado[
        [
            "total_sales",
            "na_sales",
            "jp_sales",
            "other_sales"
        ]
    ].isnull().sum()
)

print("\nValores ausentes DEPOIS:")

print(
    df_vendas_geral[
        [
            "total_sales_final",
            "na_sales_final",
            "jp_sales_final",
            "other_sales_final"
        ]
    ].isnull().sum()
)

# ============================================================
# 8 RENOMEIA AS COLUNAS PARA EXPORTAÇÃO FINAL 
# ============================================================
print(df_vendas_geral.columns.tolist())

#COPIA AS MODIFICAÇÕES PARA UMA NOVA TABELA (MANTEM A ORIGINAL) 
df_tabela_geral_vendas_2024 = df_vendas_geral[
    [
        "game_name",
        "console_code",
        "genre",
        "publisher",
        "developer",
        "critic_score",
        "total_sales_final",
        "na_sales_final",
        "jp_sales_final",
        "pal_sales",
        "other_sales_final",
        "game_release_date",
        "last_update_date"
    ]
].copy()

#RENOMEIA AS COLUNAS DE CÁLCULOS DE MÉDIA
df_tabela_geral_vendas_2024 = (
    df_tabela_geral_vendas_2024.rename(
        columns={
            "total_sales_final": "total_sales",
            "na_sales_final": "na_sales",
            "jp_sales_final": "jp_sales",
            "other_sales_final": "other_sales"
        }
    )
)
#padroniza os nomes em CONSOLE_CODE
df_tabela_geral_vendas_2024["console_code"] = (
    df_tabela_geral_vendas_2024["console_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

#JÁ CRIA UMA NOVA COLUNA COM UM ID_GAME DEFINITIVO PARA O CSV 
df_tabela_geral_vendas_2024.insert(
    0,
    "id_game",
    range(
        1,
        len(df_tabela_geral_vendas_2024) + 1
    ) # AQUI INCLUÍ 1 A CADA LINHA DA TABELA, ASSIM, CADA JOGO POSSUIRÁ UM ID_GAME ÚNICO PARA ELE;
)
```

BLOCO DE CHECAGEM FINAL DA TABELA CRIADA: 

Objetivo: VErifica se as tabelas criadas estão funcionando bem e com os dados bem estruturados


```python
# ============================================================
# 9. CHECAGEM FINAL DA TABELA: 
# ============================================================

print("\nColunas da tabela geral:")
print(
    df_tabela_geral_vendas_2024.columns.tolist()
)
print("\nPrimeiras linhas:")
print(
    df_tabela_geral_vendas_2024.head()
)
print("\nQuantidade de linhas e colunas:")
print(
    df_tabela_geral_vendas_2024.shape
)

print("\nValores ausentes na tabela geral:")

print(
    df_tabela_geral_vendas_2024
    .isnull()
    .sum()
)

print("\nLinhas completamente duplicadas:")
duplicados_jogo_console = (
    df_tabela_geral_vendas_2024
    .duplicated(
        subset=[
            "game_name",
            "console_code"
        ],
        keep=False
    )
)


print("\nRegistros repetidos por jogo e console:")

print(
    df_tabela_geral_vendas_2024[
        duplicados_jogo_console
    ][
        [
            "id_game",
            "game_name",
            "console_code",
            "total_sales"
        ]
    ]
)

# ============================================================
# 10 VERIFICA OS CONSOLE_CODE DAS TABELAS: CONSOLE E VENDAS
# ============================================================
codigos_usados_nas_vendas = set(
    df_tabela_geral_vendas_2024["console_code"]
    .dropna()
    .astype("string")
    .str.strip()
    .str.upper()
    .unique()
)

codigos_cadastrados_nos_consoles = set(
    df_consoles_sql["console_code"]
    .dropna()
    .astype("string")
    .str.strip()
    .str.upper()
    .unique()
)

codigos_sem_correspondencia = (
    codigos_usados_nas_vendas
    - codigos_cadastrados_nos_consoles
)

print(
    "\nCódigos da tabela de vendas "
    "sem console cadastrado:"
)

print(
    sorted(codigos_sem_correspondencia)
)


# ============================================================
# 11 VERIFICA A TABELA DE CONSOLES APÓS TRATAMENTO
# ============================================================

print("\nCSV DE CONSOLES PREPARADO PARA O SQL:")
print(df_consoles_sql.head())


print("\nQuantidade de linhas e colunas:")
print(df_consoles_sql.shape)


print("\nCabeçalhos finais:")
print(df_consoles_sql.columns.tolist())


print("\nTipos das colunas:")
df_consoles_sql.info()


print("\nValores ausentes:")
print(df_consoles_sql.isnull().sum())


print("\nCódigos de console duplicados:")
print(
    df_consoles_sql.duplicated(
        subset=["console_code"]
    ).sum()
)


print("\n", "=" * 80)
```
